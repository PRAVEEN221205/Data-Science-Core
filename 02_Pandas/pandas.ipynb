## Pandas Series

A Series is a one-dimensional labeled array capable of holding data of any type.


```python
data=[1,2,3,4,5]
series=pd.Series(data)
print(series)
print(type(series))
```

```python
##we can custom the index
labels=['a','b','c','d','e']
pd.Series(data,index=labels)
```

```python
##creating series with array
data=np.array([4,5,6,7,8])
pd.Series(data)
```

```python
##Creating series from dictionary elements
data={'a':1,'b':2,'c':3}
series=pd.Series(data)
print(series)
```

## Pandas DataFrame

A DataFrame is a two-dimensional labeled data structure with rows and columns.

```python
##dataframe from dictionary of lists

data={
    'Name':['Harsha','satwik','Eshwar'],
    'Age':[20,20,19],
    'City':['Srikalahasti','Hyderabad','Kurnool']
}
df=pd.DataFrame(data)
print(df)
print(type(df))
```

```python
##creating a DataFrame from list
list=[
    ['John',12,6000],
    ['Sam',15,7000],
    ['Bob',14,1300]
]
columns=['Name','Age','Salary']
df=pd.DataFrame(list,columns=columns)
df
```

```python
## create a dataframe from list of dictionary
data=[
    {'Name':'Harsha','Age':20,'City':'Srikalahasti'},
    {'Name':'Satwik','Age':20,'City':'Hyderabad'},
    {'Name':'Eshwar','Age':19,'City':'Kurnool'}
]
df=pd.DataFrame(data)
print(df)
type(df)
```

## Merging

```python
import pandas as pd

# DataFrame 1: Student information
students = pd.DataFrame({
    'student_id': [101, 102, 103, 104, 105, 106],
    'name': ['Amit', 'Neha', 'Rohit', 'Kavya', 'Vikas', 'Sara'],
    'course': ['AI', 'Data Science', 'Cyber Security', 'AI', 'Cloud Computing', 'Data Science']
})

print(students)


# DataFrame 2: Marks information
marks = pd.DataFrame({
    'student_id': [101, 102, 104, 105, 107],
    'math': [85, 78, 88, 76, 90],
    'python': [92, 81, 91, 80, 89],
    'statistics': [79, 85, 84, 82, 88]
})

print(marks)
```

```python
pd.merge(students,marks,on='student_id',how='outer')
```

```python
pd.merge(students,marks,on='student_id',how='left')
```

```python
pd.merge(students,marks,on='student_id',how='right')
```

## Group By

groupby() 

– Definition

groupby() is a Pandas function used to split the data into groups based on one or more columns so that you can perform operations on each group separately.

```python
import pandas as pd

data = {
    'Category': ['Electronics', 'Clothing', 'Electronics', 'Furniture', 'Clothing', 'Electronics', 'Furniture', 'Clothing'],
    'Store': ['S1', 'S1', 'S2', 'S1', 'S2', 'S3', 'S2', 'S3'],
    'Sales': [500, 300, 450, 700, 350, 600, 650, 400],
    'Quantity': [5, 3, 4, 7, 3, 6, 6, 4],
    'Date': pd.date_range('2023-02-01', periods=8)
}

df = pd.DataFrame(data)
df
```

```python
## group by category and find the sum of the sales
df.groupby('Category')['Sales'].sum()
```

```python
## group by Store and find the sum of the sales
df.groupby('Store')['Sales'].sum()
```

```python
## group by multiple columns
## group by Category and Store
df.groupby(['Category','Store'])['Sales'].sum()
```

## pivoit table

A pivot table is a data summarization tool used to organize, group, and aggregate data in order to analyze relationships and patterns in a dataset.

```python
import pandas as pd
import numpy as np

data = {
    'Date': pd.date_range("2023-03-01", periods=20),
    'Product': ['Laptop', 'Phone', 'Tablet', 'Laptop', 'Phone',
                'Tablet', 'Laptop', 'Phone', 'Tablet', 'Laptop',
                'Phone', 'Tablet', 'Laptop', 'Phone', 'Tablet',
                'Laptop', 'Phone', 'Tablet', 'Laptop', 'Phone'],
    
    'Region': ['North', 'South', 'East', 'West', 'North',
               'South', 'East', 'West', 'North', 'South',
               'East', 'West', 'North', 'South', 'East',
               'West', 'North', 'South', 'East', 'West'],
    
    'Sales': np.random.randint(500, 5000, 20),
    'Units': np.random.randint(1, 50, 20),
    
    'Salesperson': ['Ravi', 'Priya', 'Amit', 'Neha', 'Ravi',
                    'Priya', 'Amit', 'Neha', 'Ravi', 'Priya',
                    'Amit', 'Neha', 'Ravi', 'Priya', 'Amit',
                    'Neha', 'Ravi', 'Priya', 'Amit', 'Neha']
}

df = pd.DataFrame(data)

df['Month'] = df['Date'].dt.month_name()
df['Quarter'] = "Q" + df['Date'].dt.quarter.astype(str)

df
```

```python
pd.pivot_table(df,values='Sales',index='Region',columns='Product',aggfunc='median')
```

```python
pd.pivot_table(df,values=['Sales','Units'],index='Region',columns='Product')
```

Pivot tables are mainly designed for numeric columns

Cross Table

```python
pd.crosstab(df['Region'],df['Product'],)
```
