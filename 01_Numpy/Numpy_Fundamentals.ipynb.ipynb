| `ravel()` | Flatten array (view) |

If you want to reshape it should have exact length of the array mentioned above

Array Generating Functions

| Function           | Purpose                |

| `zeros()`          | All zeros              |

1️⃣ np.zeros() – All zeros
Creates an array filled with 0.

```python
#for 1d
print(np.zeros(2))
#for 2d we use (()) because  it is an array inside another array
print(np.zeros((2,3)))
```

2️⃣ np.ones() –Creates an array filled with 1.

```python
#for 1d
print(np.ones(4))
#for 2d
print(np.ones((3,4)))
```

3️⃣ np.full() – Creates an array filled with a specific value

```python
#for 1d
print(np.full((2),7))
#for 2d
print(np.full(((3,4)),7))
```

 4️⃣ np.arange() –Creates values from start to stop (excluding stop) with a step size.

np.arange(start, stop, step)


```python
#it start from 0 to given number-1
print(np.arange(10))
```

```python
print(np.arange(1,10,3))
```

5️⃣ np.linspace() – Equal divisions

Divides a range into equal parts (includes both ends).

np.linspace(start, stop, num)

```python
l = np.linspace(1, 10, 5)
print(l)
```

6️⃣ np.eye() – Identity matrix

Creates a square matrix with:

1 on diagonal

0 elsewhere

```python
e = np.eye(3)
print(e)
```

7️⃣ np.empty() – Uninitialized array

Creates an array without initializing values.

```python
em = np.empty((2, 2))
print(em)
```

1️⃣ np.random.rand()

Universal Function

 NumPy Universal Functions (ufuncs)

Universal functions (ufuncs) are vectorized functions that operate element-wise on NumPy arrays and support broadcasting.

“NumPy ufuncs are implemented in C and use vectorized operations, reducing Python-level loops and improving cache efficiency.”

1️⃣ Mathematical (Arithmetic) ufuncs

| `add()`      | Addition       | `np.add(a, b)`      |

| `multiply()` | Multiplication | `np.multiply(a, b)` |
| `divide()`   | Division       | `np.divide(a, b)`   |
| `power()`    | Exponent       | `np.power(a, 2)`    |

🔹 General Syntax for Slicing-
array[start : stop : step]


start → index to begin (default = 0)

stop → index to end (not included)

step → jump size (default = 1)

```python
arr[1:5]
```

```python
arr[4:]
```

```python
#example of using numpy
#normalization
#to have a mean of 0 and standard deviation
data=np.array([1,2,3,4,5])
mean=np.mean(data)
sd=np.std(data)
normalized_data=(data-mean)/sd
print(f"Normalization:",normalized_data)
```

```python
#mean
mean=np.mean(data)
#median
median=np.median(data)
#mode
standard_deviation=np.std(data)
#variance
variance=np.var(data)
print(f"mean: {mean}")
print(f"median: {median}")
print(f"standard_deviation: {standard_deviation}")
print(f"variance: {variance}")
```
